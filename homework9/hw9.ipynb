# 2 Plotting a Function
Goal: Write a **function** that will plot $f(x) = x^2$ and $g(x) = \sqrt{x}$ on the same plot and return the $f(x)$ and $g(x)$ values as separate lists. Your plot needs to have the following:
- An appropriate figure size (`figsize=(10,8)` is recommended)
- Labeled x and y axis
- A title
- Different colors and linestyles for each function
- A legend
- Your domain should be $x \in [0, 10]$. *Hint: Use `np.linspace` to set a range of values for your functions.*

Your function should return:
- A list of $f(x)$ values
- A list of $g(x)$ values.

```python
import numpy as np
import matplotlib.pyplot as plt

def plot_functions():
    """
    This function plots f(x) = x^2 and g(x) = sqrt(x) on the same plot.
    It also returns the f(x) and g(x) values as separate lists.
    
    Inputs: None
    Outputs: (f_values, g_values) - Lists of values for f(x) and g(x)
    """
    x = np.linspace(0, 10, 100)
    
    f_x = x**2  # f(x) = x^2
    g_x = np.sqrt(x)  # g(x) = sqrt(x)
    
    fig = plt.figure(figsize=(10, 8))
    
    plt.plot(x, f_x, color='red', linestyle='-', label='f(x) = x^2')  # Line plot for f(x)
    
    plt.plot(x, g_x, color='blue', linestyle='--', label='g(x) = sqrt(x)')  # Line plot for g(x)
    
    plt.title("Plot of f(x) = x^2 and g(x) = sqrt(x)", fontsize=16)
    
    plt.xlabel("x", fontsize=14)
    
    plt.ylabel("y", fontsize=14)
    
    plt.legend(loc='upper left', bbox_to_anchor=(1, 1))
    
    plt.show()
    
    # Return the values for f(x) and g(x)
    return f_x.tolist(), g_x.tolist()

f_values, g_values = plot_functions()
```

```python
# Your code here
import numpy as np
import matplotlib.pyplot as plt

def plot_functions_with_sum():
    """
    This function plots f(x) = x^2 and g(x) = sqrt(x) on the same plot.
    It also returns the f(x) and g(x) values as separate lists, along with their sums.
    
    Inputs: None
    Outputs: (f_values, g_values, sum_f, sum_g) - Lists of values for f(x) and g(x), and their sums.
    """
    x = np.linspace(0, 10, 100)
    
    f_x = x**2  # f(x) = x^2
    g_x = np.sqrt(x)  # g(x) = sqrt(x)
    
    fig = plt.figure(figsize=(10, 8))
    
    plt.plot(x, f_x, color='red', linestyle='-', label='f(x) = x^2')  # Line plot for f(x)
    
    # Plot g(x) with blue color and dashed linestyle, add label for legend
    plt.plot(x, g_x, color='blue', linestyle='--', label='g(x) = sqrt(x)')  # Line plot for g(x)
    
    plt.title("Plot of f(x) = x^2 and g(x) = sqrt(x)", fontsize=16)
    
    plt.xlabel("x", fontsize=14)
    
    plt.ylabel("y", fontsize=14)
    
    plt.legend(loc='upper left', bbox_to_anchor=(1, 1))
    
    plt.show()
    
    # Calculate the sums of f(x) and g(x)
    sum_f = sum(f_x)
    sum_g = sum(g_x)
    
    # Return the values for f(x) and g(x), along with their sums
    return f_x.tolist(), g_x.tolist(), sum_f, sum_g

f_values, g_values, sum_f, sum_g = plot_functions_with_sum()

```
